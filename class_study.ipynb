```python
# 부모 클래스 Base Class, Super Class, Parent Class
class Animal:
    # 공통 기능
    def breath(self):
        print("숨을 쉽니다.")

# 상속 (Inheritance)
class Dog(Animal):
    # 상속 받은 클래스에서만 있는 고유 기능
    def bark(self):
        print("멍멍")

# 인스턴스 생성
dog1 = Dog()    # Dog() : 설계도 / dog1 : 도면을 현재 가지고 있는 인스턴스
dog1.breath()   # 부모 클래스로부터 물려받은 메서드를 그대로 사용 가능
dog1.bark()     # 자식 클래스의  고유 기능
```
